## Creacion de datos

Un dataframe es una tabla. Contiene un array de entradas individuales, cada una con un valor especifivo.
Cada entrada corresponde a una fila y una columna.

```python
pd.DataFrame({'Yes': [50, 21], 'No': [131, 2]})
```

Los dataframe pueden tener diferentes tipos de datos, como por ejemplo cadenas de texto.

```python
pd.DataFrame({'Bob': ['I liked it.', 'It was awful.'], 'Sue': ['Pretty good.', 'Bland.']})
```

Utilizamos el pd.DataFrame(), es un constructor que genera los objetos dataframe. La forma de declarar uno nuevo es utilizando un diccionario que sus llaves son el nombre de la columna y sus valores son una lista de entradas (filas).

Normalmente esta indexado de manera ascendente (0,1,2,3,...), pero tambien se puede asignar manualmente sus valores de indice, utilizando index.

```python
pd.DataFrame({'Bob': ['I liked it.', 'It was awful.'], 
              'Sue': ['Pretty good.', 'Bland.']},
             index=['Product A', 'Product B'])
```

### Series

Una series en pandas, es una secuencia de valores de datos, si el dataframe es una tabla, una series es una lista.

Se puede crear una series con nada mas que una lista.

```python
pd.Series([1, 2, 3, 4, 5])
```

Una serie es un dataframe de una sola columna. Por lo que se puede asignar un nombre a las filas como se vio en las secciones anteriores, utilizando index. Una series no tiene un nombre de columna, tiene un nombre sobre todo.

```python
pd.Series([30, 35, 40], index=['2015 Sales', '2016 Sales', '2017 Sales'], name='Product A')
```

## Leyendo datos

La mayor parte del tiempo utilizaremos datos que ya existen. 

Los datos se guardan de muchos formatos y formas. El mas basico es el formato CSV. Cuando se abre un documento CSV se ve algo como esto:

- Product A,Product B,Product C,
- 30,21,9,
- 35,34,1,
- 41,11,11

Por lo que un CSV es un formato que separa datos por comas (Comma separated values)

La forma de leer un csv es con pd.read_csv(), ingresando la direccion donde se encuentra el documento o un link para descargar el documento de alguna pagina.

```python
path = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DA0101EN-Coursera/laptop_pricing_dataset_mod2.csv"
df = pd.read_csv(path)
df.head()
```

## Indexing y seleccion de datos

Para ver cierta cantidad de datos o datos mas especificos se utiliza df.iloc[], que devuelve todas las filas que tienen las mismas etiquetas que se entrega y devuelve un dataframe con el resultado el cual se puede guardar en otra variable para su uso.

En el ejemplo estamos buscando todas las filas que tienen una categoria numero 4 y elegimos mostrar solo las primeras 5 filas.

```python
df_loc = df.loc[df['Category']==4]
df_loc.head()
```

Se pueden agregar mas etiquetas para una busqueda mas especifica, como en el ejemplo que estamos buscando todas las filas que contengan Category 4 y Manufacturer Dell.

```python
df_loc2 = df.loc[(df['Category']==4) & (df['Manufacturer']=='Dell')]
df_loc2.head()
```

Se pueden utilizar etiquetas enteras si el indice del dataframe consta de numeros enteros

```python
df.loc[0:5, ['GPU', 'OS']]
```

Tambien se puede utilizar df.iloc[], que utiliza las posiciones enteras y ademas puede aceptar una matriz booleana.

Se puede obtener una celda en especifico del dataframe: df.iloc[0,0]

Asimismo se puede obtener un segmento de todo el dataframe: df.iloc[0:4, 2:5]

De las cuales primero se especifica la fila o cantidad de filas y la columna o cantidad de columnas.


```python
df.iloc[0:4, 2:5]
```

**¿En qué se diferencian .iloc[] y .loc[]?**

En pandas en Python, la .iloc[] propiedad accede a un grupo de filas y columnas de DataFrame mediante indexación basada en la ubicación de enteros (como df.iloc[0,0]) o matriz booleana.

La df.loc[] propiedad accede a las filas y columnas de DataFrame utilizando sus etiquetas (como df.loc[‘row_1’, ‘column_1’]) o una matriz booleana.

**¿Qué significan las siglas LOC en Python?**

En la biblioteca pandas de Python, "loc" .loc[] significa "ubicación" 

"iloc" .iloc[] significa "ubicación entera". Esto hace referencia al tipo de indexación que utiliza cada propiedad para acceder a las filas y columnas de un DataFrame.

## Filtros Booleanos

Se construye una mascara booleana (serie de True/False) y se usa apara filtrar dilas.

```python
df = pd.DataFrame({
    'producto': ['A', 'B', 'C', 'D'],
    'precio': [100, 200, 150, 300],
    'stock': [10, 5, 8, 2]
}, index=['x', 'y', 'z', 'w'])
```

De la siguiente manera solo se muestra que filas cumplen con el filtro, pero solo mostrando el valor booleano

```python
mascara = df['precio'] > 150
mascara
```

De la siguiente manera se muestra un dataframe con las filas que cumplen con el filtro

```python
mascara = df[df['precio'] > 150]
mascara
```

Se pueden añadir mas de un filtro, con pandas se usan & (y), | (o), ~ (no) — no and/or/not de Python — y cada condición va entre paréntesis:

Ejemplo con & (and):

```python
df[(df['precio'] > 100) & (df['stock'] < 10)]
```

Ejemplo con | (or):

```python
df[(df['producto'] == 'A') | (df['producto'] == 'C')]
```

Ejemplo con ~ (not):

```python
df[~(df['stock'] < 5)]
```

Se utiliza .isin() para múltiples valores

```python
df[df['producto'].isin(['A', 'C'])]
```

Se pueden mezclar filtros booleanos con loc

```python
df.loc[df['precio'] > 150, 'producto']
```

```python
df.loc[df['precio'] > 150, ['producto', 'stock']]
```

## Ordenamiento

sort_values() - Ordena por el valor de una o mas columnas

```python
df.sort_values('precio')                          # ascendente (default)
```

```python
df.sort_values('precio', ascending=False)          # descendente
```

```python
df.sort_values(['stock', 'precio'])                 # múltiples columnas
```

```python
df.sort_values(['stock', 'precio'], ascending=[True, False])  # orden distinto por columna
```

sort_index() - Ordena por el índice (nombre de fila), no por los valores:

```python
df.sort_index()
```

```python
df.sort_index(ascending=False)
```

Ambos aceptan inplace=True si quieres modificar el DataFrame directamente en vez de devolver uno nuevo 

(aunque en general es más seguro reasignar: df = df.sort_values(...)).

```python
# Top 5 productos con precio más alto
df.sort_values('precio', ascending=False).head()
```

```python
# Filtrar solo categoría "producto" con precio > 200
df.loc[(df['producto'].isin(['D', 'B'])) & (df['precio'] > 180)]
```

```python
# Seleccionar solo columnas relevantes de las primeras 5 filas
df.iloc[:5, [0, 2]]
```

Utilizando .query()

```python
df.query('precio > 180')
```

```python
df.query('producto in ("D","B") and precio > 180')
```
