# Visualización: Matplotlib y Seaborn

## Matplotlib

Es una libreria de Python para crear visualizaciones estaticas, interactivas y animadas con datos. Provee graficos flexibles y personablizables que ayuda a entender los patrones en los datos, tendencias y relaciones efectivas.

```python
import matplotlib.pyplot as plt

x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]

plt.plot(x, y)
plt.show
```

### Componentes y partes de la figura Matplotlib
**Anatomia de un grafico Matplotlib:** 

![Anatomía de una figura en Matplotlib](imagenes\anatomia_figura.png)

Las partes de una figura de Matplotlib incluyen:
- Figure (Figura): El contenedor general que aloja todos los elementos del grafico, actuando como lienzo para las visualizaciones.
- Axes (Ejes): Las areas dentro de la figura donde se grafican los datos, cada figura puede contrner multiple "axes".
- Axis (eje): Representa el eje X y el eje Y, definiendo limites, ubicacion de las marcas (ticks) y etiquetas para la interpretacion de los datos.
- Lines and Markers: El titulo da contexto al grafico, mientras que las etiquetas de los ejes describen que datos se representan en cada uno.

### Matplotlib Pyplot

Es un modulo de Matplotlib que proporciona una interfaz similar a la de MATLAB para crear graficos. Simplicgica el proceso de añadir elementos graficos, como lineas, imagenes y texto a los ejes de la figura actual.

Pasos para utilizar Pyplot: 
- Importar Matplotlib
- Crear datos: Preparar los datos en forma de array o lista.
- Graficar datos: Se usa plt.plot().
- Personalizar grafico: Añadir titulo, etiqueta a los ejes y otros elementos utilizando metodos como plt.title(), plt.xlabel(), plt.ylabel().
- Mostrar grafico: Se usa plt.show().

```python
import matplotlib.pyplot as plt

x = [0, 2, 4, 6, 8]
y = [0, 4, 16, 36, 64]
fig, ax = plt.subplots()
ax.plot(x, y, marker='o', label = 'Data Points')

ax.set_title('Basic Components of Matplotlib Figure')
ax.set_xlabel('X_Axis')
ax.set_ylabel('Y_Axis')

ax.legend()
plt.show()
```

### Diferentes tipos de graficos en Matplotlib

#### Line Chart
Es uno de los graficos mas basicos y pueden ser creados con la funcion plot(). Se utiliza para representar relacion entre dos datos, X e Y, en ejes diferente.

Los parametros x, y son coordenadas para los puntos de datos.

```python
import matplotlib.pyplot as plt

x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

plt.plot(x, y)
plt.title('Line Chart')
plt.ylabel('Y_Axis')
plt.xlabel('X_Axis')
plt.show()
```

#### Bar Chart
Grafica datos categoricos usando barras rectangulares y su largo es proporcional a los valores que representa. Puede ser graficada de forma vertical o horizintal para comparar diferentes categorias.

Parametros: 
- x: Categorias o posiciones para el eje x.
- height (altura): Altura de las barras para los valores del eje y.

```python
import matplotlib.pyplot as plt

x = ['Thur', 'Fri', 'Sat', 'Sun']
y = [170, 120, 250, 190]

plt.bar(x, y)
plt.title("Bar Chart")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

#### Histograma
Muestra la distribucion de los datos, agrupando los valores en intervalos. Se utiliza la funcion hist() para crearlo, con el eje x mostrando los intervalos y el eje y mostrando la frecuencia.

Parametros: 
- x : datos
- bins (intervalo): Numero de intervalos para agrupar los datos. 

```python
import matplotlib.pyplot as plt

x = [7, 8, 9, 10, 10, 12, 12, 12, 13, 14, 14, 15, 16, 16, 17, 18, 18, 19, 20, 20,
     21, 22, 23, 24, 25, 25, 26, 28, 30, 32, 35, 36, 38, 40, 42, 44, 48, 50]

plt.hist(x, bins=10, color='steelblue')
plt.title('Histogram')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

#### Scatter Plot
Son utilizados para observar relaciones entre variables. Se crea utilizando el metodo scatter().

Parametros x, y son las coordenadas de los puntos.

```python
import matplotlib.pyplot as plt

x = [10, 15, 20, 25, 30]
y = [12, 18, 25, 28, 35]

plt.scatter(x, y)
plt.title('Scatter Plot')
plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.show()
```

#### Pie Chart
Es un grafico circular usado para mostrar la porporcion en porcentaje de los datos. Se crea utilizando pie(). Donde cada trozo representa una parte del todo.

Parametros: 
- x: Valores de datos para los trozos del grafico.
- label (etiquetas): Nombre de cada trozo.
- autopct: El formato en que se muestran los porcentaje ('%1.1f%%')

```python
import matplotlib.pyplot as plt

cars = ['AUDI', 'BMW', 'FORD','TESLA', 'JAGUAR',]
data = [23, 10, 35, 15, 12]

plt.pie(data, labels=cars, autopct= '%1.1f%%')
plt.title('Pie Chart')
plt.show()
```

#### Box Plot
Es un grafico simple que muestra como los datos se propagan. Muestra el minimo, maximo, promedio y cuartiles, tambien ayuda a mostrar valores atipicos.

Parametros: 
- x: Datos con los que el grafico sera dibujado.
- notch: si es True la caja se dibuja con una muesca alrededor de la mediana en vez de una linea recta. Representa el intervalo de confianza de la media.
- vert: si es True el dibujo de la caja es vertival para valores del eje y, si es False dibuja una caja horizontal para valores del eje X.

```python
import matplotlib.pyplot as plt

data = [ [10, 12, 14, 15, 18, 20, 22],
         [8, 9, 11, 13, 17, 19, 21],
         [14, 16, 18, 20, 23, 25, 27] ]

plt.boxplot(data)
plt.xlabel('Groups')
plt.ylabel('Values')
plt.title('Box Plot')
plt.show()
```

#### Heatmap
Es una representacion grafica de los datos donde el valor es mostrado como un color. Ayuda a visualizar patrones, correlaciones o intensidad en un formato similar a una matriz. Se crea utilizando el metodo imshow().

Parametros:
- x: Array 2D.
- cmap: Asigna el mapa de calor.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)
data = np.random.rand(10,10)

plt.imshow(data, cmap='viridis', interpolation='nearest')

plt.colorbar()
plt.xlabel('X_Axis')
plt.ylabel('Y_Axis')
plt.title('Heatmap')
plt.show()
```

#### Características principales de Matplotlib
- **Graficado versátil:** Crea una amplia variedad de visualizaciones, incluyendo gráficos de línea, gráficos de dispersión, gráficos de barras e histogramas.
- **Personalización extensa:** Controla cada aspecto de tus gráficos, desde colores y marcadores hasta etiquetas y anotaciones.
- **Integración perfecta con NumPy:** Grafica arrays de datos directamente sin esfuerzo, potenciando las capacidades de manipulación de datos.
- **Gráficos de alta calidad:** Genera gráficos listos para publicación, con control preciso sobre la estética.
- **Compatibilidad multiplataforma:** Usa Matplotlib en Windows, macOS y Linux sin problemas.
- **Visualizaciones interactivas:** Interactúa con tus datos de forma dinámica gracias a las funciones de graficado interactivo.

## Seaborn

Es una libreria de Python para crear graficos atractivos de satadicticas. Construido en Matplotlib y integrado con Pandas, simpligica complejos graficos como graficos de linea, mapa de calor y graficos de violin con una cantidad minima de codigo. 

### Line Plot
Muestra la relacion entre dos variables numericas, a menudo con el tiempo. Tambien puede comparar multiple grupos utilizando lineas diferentes.

Parametros: 
- x, y: Variables de entrada numericos. Pueden ser array, listas o nombres de columnas de un DataFrame.
- data: Dataframe contenedor de los datos.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.lineplot(x=df.index, y='Age', data=df)
plt.xlabel('Index')
plt.ylabel('Age')
plt.title('Age Line Plot')
plt.show()
```

### Scatter Plot
Son utilizados para visualizar la relacion entre dos variables numericas. Ayudan a identificar correlaciones o patrones. Puede dibujar un grafico de dos dimensiones.

Parametros: 
- x, y: Datos de entrada que deberian ser numericos.
- data (opcional): Dataset que contiene las variables.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.scatterplot(x=df.index, y='Age', data=df)
plt.show()
```

### Box Plot
Es una representacion visual que muestra grupos de datos numericos con sus respectivos cuartiles, en relacion con datos continuos o categoricos.

Consta de cinco estadisticas claves: 
- Minimo
- Primer cuartil (25%)
- Mediana (segundo cuartil 50%)
- Tercer cuartil (75%)
- Maximo

Wide-form (formato ancho): Cada columna es una variable/categoría distinta, con sus propios valores:

-     cool  warm  hot
-     45.2  38.1  30.5
-     47.8  40.2  32.1

Long-form (formato largo): Una sola columna contiene los valores, y otra columna indica a qué categoría pertenece cada valor:

-     categoria   humedad
-     cool        45.2
-     cool        47.8
-     warm        38.1
-     warm        40.2

Parametros: 
- x, y, hue: 
- data: Dataset a graficar. Si no se especifican x e y, los datos se tratan como wide-form.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 45]} 
df = pd.DataFrame(data)

sns.boxplot(y='Age', data=df)
plt.show()
```

### Violin plot
Es similar a bloxplot. Muestra varios datos cuantitativos con funcion de una o mas variables categoricas, de modo que dicha distribuciones puedan compararse.

- La silueta ("cuerpo" del violín) representa la densidad de los datos en cada valor posible del eje Y — dónde hay muchos datos, el violín es "ancho"; donde hay pocos, es "angosto". Es esencialmente un histograma suavizado, girado verticalmente y espejado en ambos lados.
- Dentro del violín suele venir dibujada una mini "caja" tipo boxplot (mediana, cuartiles), combinando ambos tipos de información en un solo gráfico.

Parametros: 
- x, y, hue: Ingresos para graficar datos de manera long-form.
- data: Dataset para graficar.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.violinplot(y='Age', data=df)
plt.show()
```

### Swarm Plot

Muestra cada punto de datos individual sin superposición, a lo largo de un eje categórico, permitiendo ver claramente la densidad de la distribución en cada categoría. Cada punto representa una observación real (no un resumen estadístico como en el boxplot), y los puntos se "acomodan" horizontalmente para no pisarse entre sí, dejando ver dónde se concentran más o menos datos.

Parametros:
- x, y, hue: Ingresos para graficar datos de manera long-form.
- data: Dataset para graficar.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.swarmplot(x=df.index, y='Age', data=df)
plt.show()
```

### Bar Plot 
Representa una estimación de la tendencia central de una variable numérica mediante la altura de cada barra, y muestra el grado de incertidumbre alrededor de esa estimación usando barras de error. Por defecto grafica la media de cada categoría, con las barras de error indicando el intervalo de confianza.

Parametros: 
- x, y: Variables o nombre de columnas para datos long-form.
- hue: (opcional) columna para la codificacion de color.
- data: (opcional) Dataset para graficar.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'], 'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.barplot(x='Name', y='Age', data=df)
plt.show()
```

### Point Plot
Muestra estimaciones puntuales e intervalos de confianza mediante marcadores tipo scatter, representando la tendencia central de una variable numérica. A diferencia del bar plot, no usa barras sino puntos conectados por líneas, lo que facilita comparar visualmente cómo cambia esa tendencia central entre distintas categorías.

Parametros: 
- x, y: Variables o nombre de columnas para datos long-form.
- hue: (opcional) columna para la codificacion de color.
- data: (opcional) Dataset para graficar.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'JAYAN', 'ANURAG'],'Age': [21, 23, 20, 24]}
df = pd.DataFrame(data)

sns.pointplot(x='Name', y='Age', data=df)
plt.show()
```

### Count plot
Un count plot muestra la cantidad de ocurrencias de cada categoría usando barras, permitiendo visualizar la distribución de variables categóricas. A diferencia del bar plot (que resume una variable numérica por categoría), el count plot simplemente cuenta cuántas filas hay de cada valor categórico, sin necesitar una variable numérica adicional en el eje Y.

Parametros: 
- x, y: Variables o nombre de columnas para datos long-form.
- hue: (opcional) columna para la codificacion de color.
- data: (opcional) Dataset para graficar.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {'Name': ['ANSH', 'SAHIL', 'ANSH', 'JAYAN', 'ANURAG', 'ANURAG', 'ANURAG', 'SAHIL']}
df = pd.DataFrame(data)

sns.countplot(x='Name', data=df)
plt.title("Frequency of Names")
plt.show()
```

### KDE Plot
(Kernel Density Estimate) se usa para visualizar la densidad de probabilidad de una variable continua en distintos valores dentro de ese rango continuo. Es como una versión suavizada de un histograma. También se pueden graficar varias muestras en un mismo gráfico, lo que ayuda a comparar sus distribuciones de forma más eficiente en una sola visualización.

Parametros: 
- x, y: Vectores o datos clave.
- vertical: Dato booleano para graficar verticalmente.
- palette: Color de paleta.
- data: Dataframe.

```python
from sklearn.datasets import load_iris
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['Species'] = iris.target
df['Species'] = df['Species'].map({0: 'Setosa', 1: 'Versicolor', 2: 'Virginica'})
sns.kdeplot(data=df[df['Species'] == 'Virginica'], x='sepal length (cm)', fill=True, label='Virginica')
plt.legend()
plt.show()
```

```python
for especie in df['Species'].unique():
    sns.kdeplot(data=df[df['Species'] == especie], x='sepal length (cm)', fill=True, label=especie)

plt.legend()
plt.title('Distribución de longitud de sépalo por especie')
plt.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Matriz de correlación')
plt.show()
```

## Visualizing Relationships and Patterns with Seaborn

### pairplot 
Crea una matriz de gráficos para mostrar relaciones entre todas las variables numéricas de un conjunto de datos. Dibuja diagramas de dispersión para cada par de variables y gráficos de distribución en la diagonal principal.

Parámetros principalesdata: 
- data: El DataFrame de Pandas que contiene tus datos numéricos.
- hue: Permite colorear los puntos según una variable categórica, facilitando la distinción de grupos o clases.
- diag_kind: Define el gráfico de la diagonal (puede ser 'hist' para histogramas o 'kde' para líneas de densidad).
- vars: Lista opcional para seleccionar únicamente un subconjunto de columnas específicas del DataFrame.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("whitegrid")
custom_palette = sns.color_palette("husl", 8)
sns.set_palette(custom_palette)

data = sns.load_dataset("iris")
sns.pairplot(data, hue="species")
plt.show()
```

### Joint Plots
Combinan un gráfico de dispersión con las distribuciones individuales de cada variable, permitiendo ver de un vistazo tanto cómo se distribuye cada variable por separado como la relación entre ambas. Típicamente muestra el scatter plot en el centro y un histograma o KDE de cada variable en los márgenes superior y derecho.

Parametros: 
- x, y: Variables a graficar.
- data: Dataset a graficar. 
- kind (tipo): Tipo de plot a mostrar (scatter, kde, reg, etc)

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = sns.load_dataset("tips")
sns.jointplot(x="total_bill", y="tip", data=data, kind="scatter", color="#008B8B")
plt.show()
```

Esto crea un gráfico de dispersión entre total_bill y tip, con histogramas de las distribuciones individuales a lo largo de los márgenes. El parámetro kind puede configurarse como 'kde' para estimaciones de densidad de kernel, o 'reg' para gráficos de regresión.

### Grid Plot
Se usan para crear múltiples subgráficos organizados en una cuadrícula. Usando FacetGrid de Seaborn, podemos visualizar cómo interactúan las variables entre distintas categorías, lo que facilita comparar grupos o condiciones dentro de nuestro dataset.

Parametros: 
- data: Dataset a graficar.
- col, row: Variables para las columnas y filas de la cuadricula (variable categorica).
- sns.scatterplot: La funcion de grafico para aplicar a cada faceta.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

plot=sns.FacetGrid(tips, col="time", row="sex")
plot.map(sns.scatterplot, "total_bill", "tip")
plt.show()
```

## Regression Plots: Visualizing Linear Relationships

Seaborn simplifica el proceso de realizar y visualizar regresiones, específicamente regresiones lineales, lo cual es importante para identificar relaciones entre variables, detectar tendencias y hacer predicciones. Soporta dos funciones principales para la visualización de regresiones:

- regplot(): Esta función grafica un scatter plot junto con el ajuste de un modelo de regresión lineal.
- lmplot(): Esta función también grafica modelos lineales, pero ofrece más flexibilidad para manejar múltiples facetas y datasets.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')

sns.regplot(x='total_bill', y='tip', data=tips, scatter_kws={'s':10}, line_kws={'color':'red'})
plt.show()
```

## Buenas prácticas de diseño en visualización de datos

Un gráfico técnicamente correcto puede seguir siendo difícil de leer. Estas son las 
prácticas que marcan la diferencia entre un gráfico "de prueba" y uno listo para 
comunicar hallazgos a otra persona:

### 1. Títulos claros y específicos
Un título debe decir qué se está mostrando, no solo el nombre genérico del gráfico. 
`"Ventas por región"` es mejor que `"Gráfico de barras"`.

### 2. Etiquetas en los ejes, siempre
Todo gráfico necesita `xlabel` e `ylabel` que digan qué representa cada eje, incluyendo 
la unidad de medida si aplica (ej. "CO (mg/m³)", no solo "CO").

### 3. Colores con propósito, no decorativos
- Usa color para **codificar información** (categorías, valores), no solo para 
  "que se vea bonito".
- Evita más de 5-6 colores distintos en un mismo gráfico — satura la lectura.
- Considera accesibilidad: paletas como `"colorblind"` de Seaborn son legibles 
  para personas con daltonismo.

### 4. Evitar el "chartjunk"
Elementos que no aportan información (rejillas excesivas, sombras 3D innecesarias, 
demasiadas anotaciones) distraen del dato real. Menos es más.

### 5. Leyenda solo cuando hace falta
Si el gráfico tiene una sola serie de datos, una leyenda es ruido innecesario. 
Solo agrégala cuando hay múltiples categorías o grupos que distinguir.

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

# Ejemplo aplicando las buenas prácticas: título claro, etiquetas, paleta accesible
sns.set_style("whitegrid")
sns.set_palette("colorblind")

plt.figure(figsize=(7,5))
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Distribución del monto total de cuenta por día de la semana")
plt.xlabel("Día de la semana")
plt.ylabel("Monto total de la cuenta (USD)")
plt.show()
```

### Comparación: mismo gráfico, sin buenas prácticas

```python
# Mismo gráfico, sin título, sin etiquetas claras, con paleta poco accesible
plt.figure(figsize=(7,5))
sns.boxplot(data=tips, x="day", y="total_bill", palette="bright")
plt.show()
```

El segundo gráfico muestra exactamente los mismos datos, pero sin contexto: no se 
sabe qué representa cada eje ni de qué trata el gráfico sin mirar el código. Esta 
es la diferencia entre un gráfico de exploración personal y uno listo para incluir 
en un reporte o presentación.
