# Introduccion a SQL

SQL (Structured Query Lenguage) es un lenguaje de programacion para gestionar datos en bases de daros relacionales. Cuenta con diversas funciones que permite a sus usuarios leer, manipular y modificar satos.

## Una base de datos

Es una coleccion organizada de datos.

Las tablas son similares a las hojas de calculo de excel. Tienen filas y columnas. Las bases de datos siempre se organizan por columnas y estas deben tener un nombre unico.

En terminos generales se organizan en esquemas.

## Instalacion y configuracion para utilizar Jupyter

**Requisitos principales:** 
- %pip install jupysql

Importacion de librerias principales para trabajar con pandas para transformar los datos de csv en base de datos sql.

Crear o cargar la base de datos con sqlite3.


```python
import sqlite3
import pandas as pd

con = sqlite3.connect("data/sql/retail.db")

df = pd.read_csv("data/retail_sales_dataset.csv")
df.to_sql("ventas", con, if_exists="replace", index=False, method="multi")
con.close()
```

El comando "%load_ext sql" inicializa la extencion "magic" de sql en el entorno jupyter, y permite ejecutar consultas SQL dentro del codigo.

El comando "%sql sqlite:///data/sql/retail.db" crea una conexion con nuetra base de datos local de SQLite.

```python
%load_ext sql
%sql sqlite:///data/sql/retail.db
```

Una vez finalizado lo anterior ya podemos realizar consultas sql a la base de datos creada con pandas que en este caso es "ventas".

Se utiliza %sql para que el interprete entienda que esa linea es una consulta sql y %%sql para que entienda que toda la celda es una consulta sql.

El caracter "*" significa todo, entonces la salida de la tabla tendra todas las columnas disponibles de ventas.

```python
# Cdigo Python
limite = 5 

# Solo esta línea es SQL (puedes pasarle variables de Python usando {{}})
resultado = %sql SELECT * FROM ventas LIMIT {{limite}} # type: ignore

# Esto vuelve a ser código Python
print(f"Se cargaron {len(resultado)} registros.")
```

```python
%%sql 
SELECT * FROM ventas LIMIT 3;
```

## Consultas basicas

### Select y From

Son necesarios para que todas las consultas funcionen ya que indican que datos columna seleccionar y en que tabla se encuentran.

En jupyter realizar consultas con columnas como "Date" debe estar escrita con comilla doble ya que es una palabra reservada y para las que tienen un espacio en su nombre igual.

Aunque lo mejor es que los espacios se cambien por guion bajo, en esta oportunidad lo dejaremos asi y en el siguiente cuaderno lo cambiaremos para trabajar de manera mas comoda.

```python
%%sql
SELECT "Date", "Customer ID", "Gender" 
FROM ventas;
```

El orden de salida de las columnas depende del orden en que aparece en la consulta.

```python
%%sql
SELECT "Customer ID", "Date", "Gender" 
FROM ventas;
```

Cuando ejecutas una consulta SQL se obtiene una tabla que no se almacena en ninguna parte a excepcion que guardes el resultado en alguna variable utilizando Python. La  consulta tampoco modifica la tabla "ventas", contendra siempre los mismos datos cada vez que se consulte.

SQL entendera los comandos los escribas con minuscula o mayuscula (select, SELECT), pero escribirlos en mayuscula es una convencion que facilita la lectura.

### Limit
Restringe la cantidad de filas que devuelve la consulta. 

```python
%%sql
SELECT "Customer ID", "Date", "Gender" 
FROM ventas
LIMIT 5
```

### Where
Se usa para filtrar los datos de la consulta.

```python
%%sql
SELECT "Customer ID", "Date", "Gender" 
FROM ventas
WHERE "Gender" = "Male"
LIMIT 5
```

```python
%%sql
SELECT "Customer ID", "Date", "Gender" 
FROM ventas
WHERE "Date" > "2023-05-06"
LIMIT 5
```

### Operadores de comparacion
Son la forma mas basica de filtrar datos:
- Igual a: **=**
- No es igual a: **<>** o **!=**
- Mas que: **>**
- Menos que: **<**
- Mayor o igual que: **>=**
- Menos o igual que: **<=**

```python
%%sql
SELECT *
FROM ventas 
WHERE "Customer ID" >='CUST005'
```

### Aritmetica
Se pueden realizar operaciones aritmeticas usanto los mismos operadores que en excel (+,-,*,/). Sin embargo solo se puede realizar entre columnas sobre valores de una fila predeterminada. Solo se pueden sumar valores de varias columnas de la misma fila utilizando "+", si se quiere sumar valores de varias filas, se utilizan funciones de agregacion que se vera mas adelante.

Por ejemplo si no existiera la columna "Total Amount" se puede generar como salida de consulta de la siguiente manera:

```python
%%sql
SELECT "Quantity", "Price per Unit", "Price per Unit" * "Quantity" AS "Total"
FROM ventas
```

Se pueden utilizar los parentesis para el orden de las operaciones

```python
%%sql
SELECT "Quantity", "Price per Unit", ("Price per Unit" + 50) * "Quantity" AS "Total (precio unitario + 50)"
FROM ventas
```

### Operadores logicos
Permiten utilizar varios operadores de comparacion en una misla columna, algunos son: 
- **LIKE**: Permite comparar valores similares.
- **IN**: Permite especificar una lista de valores.
- **BETWEEN**: Permite seleccionar filas dentro de un rango.
- **IS NULL**: Permite seleccionar filas que no contengan datos en una columna determinada.
- **AND**: Permite seleccionar unicamente filas que cumplan dos condiciones.
- **OR**: Permite seleccionar filas que cumplan cualquiera de dos condiciones.
- **NOT**: Permite seleccionar filas que no coincidan con una condicion determinada.

#### Like
Obtendremos todos los datos donde el "Customer ID" tenga un 4 al final.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Customer ID" LIKE '%4'
```

##### Comodines y ILIKE
El "%" símbolo utilizado anteriormente representa cualquier carácter o conjunto de caracteres. En este caso, "%" se denomina "comodín". En el tipo de SQL que utiliza Mode, "LIKE" distingue entre mayúsculas y minúsculas, lo que significa que la consulta anterior solo capturará coincidencias que terminen con un "4". Para ignorar las mayúsculas y minúsculas al comparar valores, puede utilizar el ILIKE comando.

Aunque es un operador especifico de PostgreSQL, es importante tenerlo en cuanta.

#### In
Permite especificar una lista de valores que se desean incluir en los resultados.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Product Category" IN ("Electronics", "Beauty")
```

#### Between
Permite seleccionar solo las filas que se encuentran dentro de un rango específico.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Date" BETWEEN "2023-11-24" and "2023-11-30"
```

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" BETWEEN 18 and 25
```

#### Is Null
Permite excluir de los resultados las filas con datos faltantes.

El dataset utilizado no contiene datos faltantes por mala suerte para este ejemplo.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" IS NULL
```

#### And
Permite seleccionar solo las filas que cumplen dos condiciones.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" < 30 and "Product Category" = "Beauty"
```

Se puede utilziar "AND" u otro operador de comparacion tantas veces como se requiera.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" < 30 
AND "Product Category" = "Beauty"
AND "Gender" LIKE "Male"
```

#### Or
Permite seleccionar filas que cumplen cualquiera de dos condiciones.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" < 30 
OR "Gender" LIKE "Male"
```

Se pueden combinar operadore utilizando "AND" y parentesis.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" < 30
AND ("Product Category" LIKE "Clothing" AND "Date" BETWEEN "2023-02-27" AND "2023-02-31")
```

#### NOT
Se puede colocar antes de cualquier instrucción condicional para seleccionar las filas para las que esa instrucción es falsa.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" < 30 
AND "Product Category" NOT LIKE "Clothing"
```

### Order by
Permite reordenar los resultados según los datos de una o más columnas.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" < 50 and "Product Category" = "Beauty"
ORDER BY "Age"
```

Los resultados aparecen ordenados alfabeticamente (A-Z), si son numericos de menos a mayor.

Se puede organizar de manera inversa con "DESC"

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" < 50 and "Product Category" = "Beauty"
ORDER BY "Age" DESC
```

#### Ordenar datos por varias columnas
Esto es especialmente útil si los datos se dividen en categorías y se desea organizar las filas por fecha, por ejemplo, pero manteniendo juntos todos los resultados dentro de una categoría determinada.

```python
%%sql
SELECT *
FROM ventas 
WHERE "Age" < 50 and "Product Category" = "Beauty"
ORDER BY "Age" DESC, "Date"
```
