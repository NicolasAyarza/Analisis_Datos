# BD a trabajar
Datos que almacenan cursos que se imparten a distintos clientes por parte de diferentes profesores y vendido por diversos comerciales.

- **PAGADO CLIENTE:** se indica si el cliente ha pagado o no el curso.
- **PAIS:** país en el que reside el cliente al que se le ha impartido el curso.
- **CIUDAD:** ciudad en la que se ha impartido el curso.
- **PAGADO COMERCIAL:** se marca si al comercial se le ha pagado o no su comisión.

## Tablas y matrices

Son las dos visualizaciones más comunes.

### Tabla
Mustra datos en formato plano, fila por fila. Cada fila es un registro individual, cada columna un campo. 


<img src="imagenes/tabla.png" width="200">

```python
# Se obtiene el mismo resultado de esta manera con pandas
df.groupby('CURSO')['DURACION'].agg(sum)
```

### Matriz
Muestra los datos agrupados jerarquicamente, con filas y columnas que se puede expandir/colapsar.

<img src="imagenes/matriz.png" width="350">

```python
resultado = df.pivot_table(
    index='CLIENTE', 
    columns='JORNADA CURSO', 
    values='DURACION', 
    aggfunc='sum', 
    fill_value=0,
    margins=True,
    margins_name="Total"
    ).head(9)

resultado.style.format('{:.0f} horas')
```

## Segmentacion de datos
Es un tipo de informe que se utiliza para filtrar todas las visualizaciones de la pagina. A ella solo se agrega un unico campo, de tal modo que al seleccionar un elemento del campo agregado, se emite un filtro por dicho elemento a todos los informes de la pagina.

## Tarjeta

Un elemento visual que admite un unico campo, la cual le aplicara una funcion de totales y mostrara su valor en la tarjeta. Es decir, en una tarjeta totalizamos un campo para que nos muestre su total.

Solo puede recibir filtros, no emitirlos.

## Graficos de barra y columna

En los gráficos de barras, estas crecen sobre el eje horizontal (X), por lo que el campo a totalizar lo colocaremos sobre dicho eje. Sin embargo, en los gráficos en columnas, estas crecen sobre el eje vertical (Y), por lo que el campo que totalizamos lo colocamos en dicho eje.

## Gráficos de columnas y líneas
Este tipo de gráfico nos presenta dos ejes verticales, uno para medir las columnas y otro para medir la línea del gráfico. Lo utilizamos cuando queremos representar en un mismo gráfico dos campos totalizados, uno con valores altos y otro con valores bajos. De esta forma, uno se medirá en un eje vertical principal y el otro en el eje vertical secundario.

## Gráfico circular
Este tipo de gráfico lo utilizamos cuando no necesitamos ejes, es decir, cuando solo queremos ver la evolución de los elementos de un solo campo. Por ejemplo, queremos ver el número de cursos que se han dado en cada jornada.

Algunas de las propiedades de formato más habituales de los gráficos son:

- COLUMNAS/BARRAS: en función de si es un gráfico de columnas o de barras, para configurar cada una de las series del gráfico.
- SEGMENTOS: si el gráfico es circular, para configurar cada serie de este.
- LEYENDA: para formatear, mover o quitar la leyenda del gráfico.
- ETIQUETAS DE DATOS: para mostrar y configurar los valores de cada una de las series del gráfico.
