## Graficos

### Jerarquias de fecha
Cada vez que se agrega un campo fecha a un grafico, se genera una jerarquia de fechas, lo que provoca que el propio campo fecha se divida en cuatro segmentos o subpartes de la fecha:
- **Año**: Muestra solo el año de cada fecha del campo
- **Trimestre**: Muestra los diferentes trimestres en los que hay fechas en cada año del campo.
- **Mes**: Muestra los diferentes meses en loa que hau fechas en dicho campo.
- **Dia**: Muestra los dias de cada mes y año en los que hau fecha en dicho campoñ.

### Grafico de lineas para tendencias de tiempo
El grafico de linea es una visualizacion muy util cuando se quiere analizar la evolucion de un valor a lo largo del tiempo o de una secuencia ordenada. Es especialmente eficaz para identificar tendencias, patrones, estacionalidades o fluctuaciones en los datos.

Se utiliza para:
- Mostrar evolucion de una metrica a lo largo del tiempo (Ventas mensuales, numero de clientes por trimestre, entre otros)
- Comparar varias series temporales en paralelo.
- Detectar subidas, bajadas o cambios de tendencia.
- Identificar periodos de estabilidad o de variacion significativa.

### Graficos para representaciones geograficas
Estos graficos permiten visualizar datos en mapas, ofreciendo una forma intuitiva y potente de analizar informacion relacionada con ubicaciones.

Se tienen diferentes visualizaciones geograficas, como los mapas o mapas coropleticos, que permiten mostrar datos por pais, region, ciudad, etc.

Se utilizan para: 
- Analizar datos relacionados con ubicaciones geograficas.
- Detectar patrones o tendencias por zona geografica (por ejemplo, ventas por comunidad autonoma).
- Realizar comparativas regionales o internacionales de forma visual.
- Identificar areas de oportunidad o zonas de baja actividad.

#### Mapa coropletico
Es una visualizacion geografica que representa datos mediante el relleno de areas geograficas, como paises, regiones, provincias. Utilizando colores o intensidades segun el valor de un campo. Permite comparar zonas geograficas de forma visual, facilitando la identificacion y distribucion de datos sobre el mapa.

### Interacciones 
Permiten controlar cómo un objeto visual afecta o se relaciona con los demás en una misma página de informe.

Tipos de interaccion: 

- Filtrar: Oculta los datos que no corresponden a la selección, mostrando únicamente la información relevante como si fuera un segmento.    
- Resaltar: Es la opción por defecto. Mantiene todos los datos visibles, pero atenúa los no seleccionados y destaca la proporción elegida.  
- Ninguno: Desconecta el gráfico. La selección en una visualización no produce ningún cambio en la otra.

### Marcadores

## Power Query

Es un motor de ETL de Power BI. Se usa antes de que los datos lleguen al modelo de datos.

1) Conexion y extraccion de datos:
- Conectar a multiples fuentes: Excel, CSV, Base de datos SQL, Carpetas, Json, ETC.
- Combinar datos de varias fuentes en una sola consulta.
- Crear parametros para hacer las conexiones dinamicas.

2) Limpieza de datos: 
- Quitar filas duplicadas.
- Cambiar tipos de datos
- Reemplazar valores, rellenar nulos, recortar espacios.
- Filtrar filas segun condiciones.
- Detectar y corregir errores con "Reemplazar errores" o "Quitar errores".

3) Transformacion de columnas:
- Dividir colimnas.
- Combinar/concatenar columnas.
- Extraer texto.
- Pivotar y despivotar columnas.
- Agregar columnas condicionales.
- Agregar columna indice.
- Redondear, calcular operaciones matematicas entre columnas.

4) Combinacion de tablas:
- Merge queries (equivalente a un JOIN).
- Append queries (Equivalente a UNION).
- Combinar consultas dentro de la misma consulta o crear una nueva consulta.

5) Agrupacion y agregacion
- Group By: agrupar filas y calcular sumas, promedios, conteos, maximos, minimos, etc.

6) Gestion de consultas:
- Ver y editar el historial de pasos aplicados, redondear o eliminar pasos.
- Duplicar o referenciar consultas para no repetir transformaciones.
- Deshabilitar la carga de consultas intermedias.
- Organizar consiltas en grupos/carpetas.

7) Actualizacion de datos:
- Configurar la actualizacion para que las transformaciones se apliquen automaticamente cada vez que se actualizan los daos, sin tener que rehacer el proceso manualmente.

### Crear columna personalizada

Los operadores que podemos utilizar son: 

- +: para sumar 
- -: para restar 
- *: para multiplicar 
- /: para dividir 
- &: para juntar texto.

## Importar una carpeta

Esta opcion permite conectar Power Bi a una carpeta que contiene varios archivos, como por ejemplo:
- .xlsx
- .csv
- TXT

El objetivo principal es combinar automaticamente varios archivos que tienen la misma estructura.

Obtiene informacion como:
- Nombre
- Extencion
- Fecha de modificacion
- Ruta
- Contenido

Se combinan los archivos mediante Power Query.

Power BI genera automaticamente elementos como: consulta principal, archivo de ejemplo, Funcion para transformar archivos, parametros.

Esto permite que, si despues de agregar otro archivo con la misma estructura a la carpeta, Power BI pueda incor´prarlo al actualizar.


## Crear modelos de datos

- **Tabla de hechos:** es la tabla que tiene toda la informacion relevante de la cual se quiere hacer el analisis con graficos.
- **Tabla de dimensiones:** es la tabla que tiene todos los campos clave que aportan informacion a la tabla de hechos.

Para poder relacionar dos tablas, ambas han de tener una columna coincidente, no tiene por que tener el mismo nombre, pero si los mismos elementos.

En la tabla de dimensiones, los elementos de la columna coincidente no se deben repetir, han de ser elementos unicos, ya que tienen que referenciar e identificar a cada uno de los valores que estan describiendo, y de esta forma, poder establecer un tipo de relacion de **uno a muchos**, para no duplicar resultados.

## Modelo estrella

Orgabiza la infirmacion en uns estructura sencilla y facil de optimizar para consultas analiticas.

Se compone de una tabla central (tabla de hechos) que almacena datos a cuantificar y en torno a la que gira todos los informes que se obtengan del modelo y varias tablas dimensiones. 

Se unen con la relacion donde la tabla de hechos siempre queda al lado de muchos, mientras que las tablas dimensiones estaran en el lado de 1.

Cada dimension se conecta a la tabla de hechos por una clave unica como por ejemplo: Codigo de producto.

### Estructura del modelo de datos estrella
Antes de comenzar a crear el modelo de datos en Power BI, es importante realizar una correcta limpieza y transformación de los datos.  

Esto incluye eliminación de datos duplicados, de valores en blanco y ajustar bien los tipos de datos, tarea que se realiza en Power Query. 

Una vez que están las tablas bien estructuradas se cargarán en Power BI para crear el modelo de datos, donde es aconsejable, tener claras las relaciones a establecer y el modelo a crear, donde se valorará: 

- Quién es la tabla de hechos y las columnas coincidentes con las tablas de dimensiones. 
- Cuáles son las tablas de dimensiones y el campo coincidente de cada una de ellas respecto a la tabla de hechos. 
- Asegurar que los elementos de las columnas coincidentes de cada una de las tablas de dimensiones no se repitan. 
- Chequear que el tipo de dato de las columnas coincidentes de la tabla de hechos con las tablas de dimensiones sea el mismo. 
- Estar seguros de que las tablas de dimensiones aportan información relevante sobre los campos clave de la tabla de hechos. 

Una vez creado el modelo de datos, a la hora de utilizar los campos en los informes, debemos de tener en cuenta que: 

- Los campos por los que se vayan a agrupar los elementos o a filtrar los datos, se agregarán de las tablas de dimensiones (siempre que se dispongan de ellos). 
- Los campos a totalizar en cualquier informe siempre se agregan de la tabla de hechos. 
- Los campos fecha para ver informes de tiempo o filtros en el tiempo se seleccionarán de la tabla de hechos.

## Calculos DAX

Es el lenguaje para crear calculos personalizados que van mas alla de las funcionalidades basicas de las tablas y graficos. Se puede obtener informacion mas precisa, realizar analisis complejos y adaptar los resultados a las necesidades especificas del informe.

Calculos a centrar: 
- **Columnas calculadas:** su valor se calcula fila por fila por fila en funcion de otras columnas o expresiones. Estas columnas se almacenan en el modelo de datos y se actualiza cada vez que se recargan los datos.
- **Medidas:** Realiza calculos dinamicos en funcion del contexto de filtrado del informe. A diferencia de las columnas calculadas, no se almacenan ficicamente en el modelo, es decir, no se guardan en ninfuna tabla, solo se aloja en ellas y se calculan sobre los informes a las que las agregamos.

Mas funciones DAX: https://learn.microsoft.com/es-es/dax/dax-function-reference

### Columnas calculadas
Se realiza un calculo entre varias columnas de una sola tabla. En caso de querer utilizar mas de una tabla, se tiene que crear realcion entre las tablas que se quieran utilizar y al momento de querer utilizar datos de otra tabla se debe utilizar la palabra **RELATED**.

Algunas funciones DAX que pueden resultar interesantes para cálculos de columnas son: 

- BLANK: devuelve un valor en blanco que se usa como equivalente a nulo en DAX. 
- ISBLANK: comprueba si un valor es en blanco y devuelve verdadero o falso. 
- DIVIDE: realiza una división entre dos valores y, a diferencia del operador de división estándar, permite especificar un valor alternativo en caso de que el divisor sea cero o esté en blanco. 
- SWITCH: evalúa una expresión y devuelve un resultado de una lista de posibles valores, simplificando estructuras condicionales múltiples. 

### Calculo de medidas
Es un calculo dinamico que se evalua en el contexto de filtrado y segmentacion que este acctivo en una visualizacion o informe.

No ocupa espacio adicional en el modelo, ya que no se almacena como dato, solo guarda la formula. Se utiliza para totalizar los datos de cualquier campo de una tabla, como sumar, contar, hacer promedio, etc.

Por lo tanto las medidas solo se calculan en los informes y no en la actualizacion de datos y fila a fila como las columnas calculadas.

Algunas funciones DAX para calculos de medidas que totalicen campos de una tabla: 
- SUM: devuelve la suma de todos los valores numéricos de una columna. 
- COUNT: devuelve el número de elementos de una columna que no están vacíos. 
- AVERAGE: devuelve el promedio aritmético de los valores numéricos de una columna. 
- MAX: devuelve el valor más alto dentro de una columna. 
- MIN: devuelve el valor más bajo dentro de una columna. 
- DISTINCTCOUNT: devuelve el número de valores únicos diferentes en una columna. 
- COUNTROWS: devuelve el número de filas de una tabla.
