 # Importacion de datos
 Se importan todos los datos de las 7 tablas de la base de datos healthcare que esta en PostgreSQL utilizando:
- load_dotenv: para buscar el .env con las claves y el url de mi base de datos.
- create_engine: crea la conexion con la base de datos para realizar la consulta y extraer todo.

Despues se realizan las consultas con pandas especificamente con read_sql, la cual permite que la consulta se transforme en un dataframe para utilizarlo mas adelante.

Al final y no menos importante se verifica si la cantidad de filas y columnas se importaron de manera correcta.

```python
from dotenv import load_dotenv 
import pandas as pd
from sqlalchemy import create_engine
import os

load_dotenv()

DB_URL = os.getenv("DATABASE_URL")

engine = create_engine(DB_URL)

df_atenciones = pd.read_sql("SELECT * FROM atenciones", con=engine)
df_citas = pd.read_sql("SELECT * FROM citas", con=engine)
df_diagnosticos = pd.read_sql("SELECT * FROM diagnosticos", con=engine)
df_especialidades = pd.read_sql("SELECT * FROM especialidades", con=engine)
df_medicos = pd.read_sql("SELECT * FROM medicos", con=engine)
df_pacientes = pd.read_sql("SELECT * FROM pacientes", con=engine)
df_servicios = pd.read_sql("SELECT * FROM servicios", con=engine)

print('Atenciones: ', df_atenciones.shape)
print('Citas: ', df_citas.shape)
print('Diagnosticos: ', df_diagnosticos.shape)
print('Especialidades: ', df_especialidades.shape)
print('Medicos: ', df_medicos.shape)
print('Pacientes: ', df_pacientes.shape)
print('Servicios: ', df_servicios.shape)
```
