## Tabla analitica unificada

```python
tb_analitica = pd.merge(df_citas, df_atenciones, on='cita_id', how='left')
tb_analitica.head(2)
```

```python
tb_analitica = pd.merge(tb_analitica, df_pacientes.rename(columns={'nombre':'nombre paciente'}), on='paciente_id', how='left')
tb_analitica.head(2)
```

```python
tb_analitica = pd.merge(tb_analitica, df_medicos.rename(columns={'nombre':'nombre medico'}), on='medico_id', how='left')
tb_analitica.head(2)
```

```python
tb_analitica = pd.merge(tb_analitica, df_especialidades, on='especialidad_id', how='left')
tb_analitica.head(2)
```

```python
tb_analitica = pd.merge(tb_analitica, df_servicios, on='servicio_id', how='left')
tb_analitica.head(2)
```

```python
tb_analitica = pd.merge(tb_analitica, df_diagnosticos, on='atencion_id', how='left')
tb_analitica.head(2)
```

```python
tb_analitica.isnull().sum()
```

```python
tb_analitica.columns
```

## Variables derivadas

```python
tb_analitica['edad paciente en cita'] = ((tb_analitica['fecha_cita'] - tb_analitica['fecha_nacimiento']).dt.days // 365.25).astype(int)
tb_analitica['edad paciente en cita'].head()
```

```python
tb_analitica['dia semana cita'] = tb_analitica['fecha_cita'].dt.day_name(locale='es_ES')
tb_analitica['dia semana cita'].head()
```

```python
tb_analitica[tb_analitica['edad paciente en cita'] < 0].shape
```
